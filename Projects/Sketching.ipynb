```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('data/pedestrians.jpg')
plt.axis('off')
plt.imshow(image[:,:,::-1]);
```

```python
def show_image(im, ax=None, figsize=(10,8)):
    if not ax: _, ax = plt.subplots(1,1,figsize=figsize)
    if len(im.shape) == 2: im = np.tile(im[:,:,None], 3)
    ax.imshow(im[:,:,::-1])
    ax.xaxis.set_visible(False)
    ax.yaxis.set_visible(False)
    return ax
```

## Oil Painting Effect

```python
def oil_effect():
    res = cv2.xphoto.oilPainting(image, 7, 1)  #
    return res

show_image(oil_effect());
```

## Watercolor Effect

```python
res = cv2.stylization(image, sigma_s=60, sigma_r=0.5)
plt.axis('off')
plt.imshow(res[:,:,::-1]);
```

## Black and White Colored Pencil Sketches

```python
# grayscale_sketch = cv2.pencilSketch(image, sigma_s=60, sigma_r=0.5, shade_factor=0.05)
# grayscale_sketch, colored_sketch = cv2.pencilSketch(image, sigma_s=100, sigma_r=0.1, shade_factor=0.02)
# plt.axis('off')
# plt.imshow(colored_sketch);
gray_img = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
inverted_img = cv2.bitwise_not(gray_img)
blurred_img = cv2.GaussianBlur(inverted_img, (21, 21), sigmaX=0, sigmaY=0)
inverted_blurred_img = cv2.bitwise_not(blurred_img)
sketch_img = cv2.divide(gray_img, inverted_blurred_img, scale=256)
plt.axis('off')
plt.imshow(sketch_img);
```
